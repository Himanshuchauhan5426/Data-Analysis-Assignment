Identify outliers in Loan Amount and EMI using a statistical method (IQR or Z-score)

```python
#Loan Amount Outliers
Q1 = df['Loan Amount (INR)'].quantile(0.25)
Q3 = df['Loan Amount (INR)'].quantile(0.75)
IQR = Q3-Q1
lower_bound = Q1-1.5*IQR
upper_bound = Q3+1.5*IQR
loan_amount_outliers = df[(df['Loan Amount (INR)']<lower_bound)|(df['Loan Amount (INR)']>upper_bound)]
print("loan_amount_outliers",len(loan_amount_outliers))
```

```python
#EMI Outliers
Q1 = df['EMI (INR)'].quantile(0.25)
Q3 = df['EMI (INR)'].quantile(0.75)
IQR = Q3-Q1
lower_bound = Q1-1.5*IQR
upper_bound = Q3+1.5*IQR
EMI_outliers = df[(df['EMI (INR)']<lower_bound)|(df['EMI (INR)']>upper_bound)]
print('EMI_outliers',len(EMI_outliers))
```

```python
#So,using the IQR method ,we identified 49 unusual loan amount values and 42 unusal EMI values
```

# Break down default rate by Loan Type, Employment Type, and Credit Score band

#default rate by loan type

```python
loan_type_default = (
    df.assign(
        Default_Flag=df['Loan Status'].isin(['Defaulted', 'Written Off'])
    )
    .groupby('Loan Type')['Default_Flag']
    .mean()
    .mul(100)
    .round(2)
)

loan_type_default
```

```python
#default rate by employment type
```

```python
employment_default = (
    df.assign(
        Default_Flag=df['Loan Status'].isin(['Defaulted', 'Written Off'])
    )
    .groupby('Employment Type')['Default_Flag']
    .mean()
    .mul(100)
    .round(2)
)

employment_default
```

```python
##default rate by Credit Score band
```

```python
credit_default = (
    df.assign(
        credit_band=pd.cut(
            df['Credit Score'],
            bins=[-float('inf'),650,750,float('inf')],
            labels=['<650','650-750','750+']
        ),
        Default_Flag=df['Loan Status'].isin(['Defaulted', 'Written Off'])
    )
    .groupby('credit_band',observed=False)['Default_Flag']
    .mean()
    .mul(100)
    .round(2)
)

credit_default
```

```python
#Create traget variable
df['Default_Flag'] = df['Loan Status'].isin(['Defaulted','Written off']).astype(int)
```

```python
#Count each class
print(df['Default_Flag'].value_counts())
```

```python
#Check the required columns
print(df[['Credit Score','EMI_to_Income_Ratio','Loan Type']].head())
```

```python
from sklearn.model_selection import train_test_split
```

```python
#select the input feature
X = df[['Credit Score','EMI_to_Income_Ratio','Loan Type']]
#select the target
y = df['Default_Flag']
```

```python
#import library for onehot coding
from sklearn.preprocessing import OneHotEncoder
```

```python
#import the model
from sklearn.linear_model import LogisticRegression
```

```python
#Numerical columns
numeric_features=['Credit Score','EMI_to_Income_Ratio']
#Categorical column
categorical_features=['Loan Type']
```

```python
encoder = OneHotEncoder(handle_unknown='ignore',sparse_output=False)
loan_type_encoded = encoder.fit_transform(X[['Loan Type']])
```

```python
X_encoded = np.column_stack([
    X[['Credit Score','EMI_to_Income_Ratio']].to_numpy(),
    loan_type_encoded])
```

```python
print(encoder.get_feature_names_out(['Loan Type']))
print(loan_type_encoded[:5])
```

```python
print(X_encoded.shape)
```

```python
#Split data into training & testing test
X_train,X_test,y_train,y_test = train_test_split(X_encoded,y,test_size=0.2,random_state=42,stratify=y)
```

```python
#Train Logistic Regression
model = LogisticRegression()
model.fit(X_train,y_train)
```

```python
#Make Predictions
y_pred = model.predict(X_test)
print(y_pred[:10])
```

```python
from sklearn.metrics import accuracy_score
```

```python
#Model Accuracy
accuracy = accuracy_score(y_test,y_pred)
print("Model Accuracy:",accuracy*100,"%")
```

```python
from sklearn.metrics import confusion_matrix
```

```python
cm = confusion_matrix(y_test,y_pred)
print(cm)
```

```python
#Classification report
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,y_pred,zero_division=0))
```
